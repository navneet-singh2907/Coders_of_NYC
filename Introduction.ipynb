```python
# Coders of the New York
with open("initialdata.txt", encoding = 'utf-8') as f:
    data = f.read()
```

```python
chunks = data.split("\n\n")
chunks = [c for c in chunks if len(c)>3]
```

```python
chunks[4]
```

```python
def parse_chunk(chunk):
    chunk = chunk.strip()
    # Variable for splitting the data with new line
    sep_chunk = chunk.split('\n')
    username = sep_chunk[0]
    no_of_posts = sep_chunk[1]
    no_of_followers = sep_chunk[2]
    no_of_following = sep_chunk[3]
    name = sep_chunk[4]
    if (len(sep_chunk) >5):
        type_of_page = sep_chunk[5]
        bio ="\n" .join(sep_chunk[6:])
    else:
        type_of_page = "unknown"
        bio = ""
    print(username,no_of_posts,no_of_followers, no_of_following,name,type_of_page,bio, sep = "\n")
    return{"username":username,"no_of_posts":no_of_posts,"no_of_followers":no_of_followers, "no_of_following":no_of_following,"name":name,"type_of_page":type_of_page,"bio":bio}

parse_chunk(chunks[0])


    
    
```

```python
all_chunks = []

for chunk in chunks:
    parsed_chunk = parse_chunk(chunk)
    all_chunks.append(parsed_chunk)

#print(all_chunks)
```
